```python
numerical_columns = [
    "CGPA",
    "Class 10 Percentage",
    "Class 12 Percentage",
    "Backlogs",
    "Number of Projects",
    "Number of Internships",
    "Number of Certifications",
    "Number of Achievements"
]

df[numerical_columns].describe()
```

```python
for column in numerical_columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower) |
        (df[column] > upper)
    ][column]

    print(column, ":", len(outliers))
```

```python
import matplotlib.pyplot as plt

df[numerical_columns].plot(
    kind="box",
    figsize=(12, 6),
    rot=45
)

plt.title("Outlier Analysis")
plt.show()
```

```python
numerical_columns = [
    "CGPA",
    "Class 10 Percentage",
    "Class 12 Percentage",
    "Backlogs",
    "Number of Projects",
    "Number of Internships",
    "Number of Certifications",
    "Number of Achievements"
]

for column in numerical_columns:

    # First recover the candidate's own known value
    candidate_value = (
        df.groupby("Candidate ID")[column]
        .transform("median")
    )

    df[column] = df[column].fillna(candidate_value)

    # If ALL applications for a candidate are missing,
    # use the global median as fallback
    df[column] = df[column].fillna(
        df[column].median()
    )
```

```python
candidate_specialization = (
    df.groupby("Candidate ID")["Specialization"]
    .transform(
        lambda x: x.mode().iloc[0]
        if not x.mode().empty
        else np.nan
    )
)

df["Specialization"] = df["Specialization"].fillna(
    candidate_specialization
)

df["Specialization"] = df["Specialization"].fillna(
    df["Specialization"].mode()[0]
)
```

```python
text_columns = [
    "Skills",
    "Projects",
    "Internships",
    "Certifications",
    "Achievements"
]

for column in text_columns:

    # Recover the candidate's known profile value
    candidate_value = (
        df.groupby("Candidate ID")[column]
        .transform(
            lambda x: x.dropna().iloc[0]
            if x.notna().any()
            else np.nan
        )
    )

    df[column] = df[column].fillna(candidate_value)

    # If the candidate has no available value anywhere,
    # mark it as Not Provided
    df[column] = df[column].fillna("Not Provided")
```
