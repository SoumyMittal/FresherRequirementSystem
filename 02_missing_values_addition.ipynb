```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("AI_Fresher_Recruitment_Dataset.csv")
```

I calculate the exact number of rows corresponding to the desired missing percentage using int(len(df) * percentage). Then I use df.sample() to randomly select that many rows without replacement. Using df.loc and np.nan, I replace the selected values with missing values. For related fields such as Projects, Internships, Certifications, and Achievements, I select the same rows and set both the count and description columns to missing. This gives controlled missingness instead of relying on a probability that only produces approximately the desired percentage.

```python
print("Rows:", len(df))

print(
    df[
        [
            "Projects",
            "Internships",
            "Certifications",
            "Achievements"
        ]
    ].isna().sum()
)
```

```python
# Specialization - exactly 2%
rows = int(len(df) * 0.02)
df.loc[df.sample(rows, random_state=1).index, "Specialization"] = np.nan


# Class 10 - exactly 2%
rows = int(len(df) * 0.02)
df.loc[df.sample(rows, random_state=2).index, "Class 10 Percentage"] = np.nan


# Class 12 - exactly 2%
rows = int(len(df) * 0.02)
df.loc[df.sample(rows, random_state=3).index, "Class 12 Percentage"] = np.nan


# Skills - exactly 3%
rows = int(len(df) * 0.03)
df.loc[df.sample(rows, random_state=4).index, "Skills"] = np.nan


# Projects - exactly 5%
eligible = df.index[
    df["Number of Projects"] > 0
]
rows = int(len(df) * 0.05)
selected = np.random.RandomState(5).choice(
    eligible,
    size=min(rows, len(eligible)),
    replace=False
)
df.loc[
    selected,
    ["Number of Projects", "Projects"]
] = np.nan


# Internships - exactly 6%
eligible = df.index[
    df["Number of Internships"] > 0
]
rows = int(len(df) * 0.06)
selected = np.random.RandomState(6).choice(
    eligible,
    size=min(rows, len(eligible)),
    replace=False
)
df.loc[
    selected,
    ["Number of Internships", "Internships"]
] = np.nan


# Certifications - exactly 7%
eligible = df.index[
    df["Number of Certifications"] > 0
]
rows = int(len(df) * 0.07)
selected = np.random.RandomState(7).choice(
    eligible,
    size=min(rows, len(eligible)),
    replace=False
)
df.loc[
    selected,
    ["Number of Certifications", "Certifications"]
] = np.nan


# Achievements - exactly 8%
eligible = df.index[
    df["Number of Achievements"] > 0
]
rows = int(len(df) * 0.08)
selected = np.random.RandomState(8).choice(
    eligible,
    size=min(rows, len(eligible)),
    replace=False
)
df.loc[
    selected,
    ["Number of Achievements", "Achievements"]
] = np.nan

# CGPA - exactly 2%
rows = int(len(df) * 0.02)
df.loc[df.sample(rows, random_state=9).index, "CGPA"] = pd.NA

# Backlogs - exactly 2%
rows = int(len(df) * 0.02)
df.loc[df.sample(rows, random_state=10).index, "Backlogs"] = pd.NA
```

```python
#Checking for missing values
df.isnull().sum()
```

```python
df.to_csv(
    "Synthetic_AI_Fresher_Recruitment_Dataset.csv",
    index=False
)
```
