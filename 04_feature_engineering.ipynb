```python
import numpy as np
import pandas as pd
```

```python
np.random.seed(42)
```

```python
df = pd.read_csv(
    "AI_Fresher_Recruitment_Dataset_Cleaned.csv"
)
```

```python
df["CGPA_Score"] = df["CGPA"] / 10
```

```python
df["CGPA_Eligible"] = (
    df["CGPA"] >= df["Minimum CGPA"]
).astype(int)
```

```python
df["CGPA_Gap"] = df["CGPA"] - df["Minimum CGPA"]
```

```python
def degree_match(row):

    if row["Degree"] in row["Preferred Degree"].split(", "):
        return 1
    return 0


df["Degree_Match"] = df.apply(
    degree_match,
    axis=1
)
```

```python
def skill_match(row):

    candidate_skills = set(
        row["Skills"].split(", ")
    )

    required_skills = set(
        row["Required Skills"].split(", ")
    )

    matched_skills = candidate_skills.intersection(
        required_skills
    )

    return round(
        len(matched_skills) / len(required_skills),
        2
    )


df["Skill_Match"] = df.apply(
    skill_match,
    axis=1
)
```

```python
df["Project_Count"] = df["Number of Projects"]

df["Internship_Count"] = df["Number of Internships"]

df["Certification_Count"] = (
    df["Number of Certifications"]
)

df["Achievement_Count"] = (
    df["Number of Achievements"]
)

df["Backlog_Score"] = 1 - (
    df["Backlogs"] / 6
)
```

```python
feature_columns = [
    "CGPA_Score",
    "CGPA_Eligible",
    "CGPA_Gap",
    "Degree_Match",
    "Skill_Match",
    "Project_Count",
    "Internship_Count",
    "Certification_Count",
    "Achievement_Count",
    "Backlog_Score"
]

df[feature_columns].head()
```

```python
df[feature_columns].describe()
```

```python
df["Project_Score"] = df["Project_Count"] / df["Project_Count"].max()

df["Internship_Score"] = (
    df["Internship_Count"] /
    df["Internship_Count"].max()
)

df["Certification_Score"] = (
    df["Certification_Count"] /
    df["Certification_Count"].max()
)

df["Achievement_Score"] = (
    df["Achievement_Count"] /
    df["Achievement_Count"].max()
)
```

```python
df["Recruitment_Score"] = (
    df["Skill_Match"] * 0.35
    + df["CGPA_Score"] * 0.20
    + df["Project_Score"] * 0.15
    + df["Internship_Score"] * 0.10
    + df["Certification_Score"] * 0.05
    + df["Achievement_Score"] * 0.05
    + df["Degree_Match"] * 0.05
    + df["Backlog_Score"] * 0.05
)
```

```python
df["Recruitment_Score"] = (
    df["Recruitment_Score"] * 100
).round(2)
```

```python
df["Recruitment_Score"].describe()
```

```python
df[
    [
        "Candidate ID",
        "Job Role",
        "Skill_Match",
        "CGPA_Score",
        "Project_Score",
        "Internship_Score",
        "Degree_Match",
        "Backlog_Score",
        "Recruitment_Score"
    ]
].head(10)
```

```python
df["Recruitment_Score"] = (
    df["Recruitment_Score"]
    + np.random.normal(0, 2, len(df))
).clip(0, 100).round(2)
```

```python
df.loc[
    df["CGPA_Eligible"] == 0,
    "Recruitment_Score"
] *= 0.70
```

```python
df.loc[
    df["Degree_Match"] == 0,
    "Recruitment_Score"
] *= 0.80
```

```python
df["Rank"] = (
    df.groupby("Job ID")["Recruitment_Score"]
    .rank(
        method="first",
        ascending=False
    )
    .astype(int)
)
```

```python
df["Shortlist Limit"] = (
    df["Number of Openings"] * 3.5
).round().astype(int)

df["Shortlisted"] = (
    df["Rank"] <= df["Shortlist Limit"]
).astype(int)
```
