```python
#Generating a realistic random list of skills for one candidate.
def generate_skills():

    number_of_skills = random.choices(
        list(skill_count_distribution.keys()),
        weights = list(skill_count_distribution.values()),
        k = 1
    )[0]

    profile = random.choices(
        list(skill_profiles.keys()),
        weights = [
            skill_profiles[p]["weight"]
            for p in skill_profiles.keys()
        ],
        k = 1
    )[0]

    available_skills = skill_profiles[profile]["skills"].copy()

    if number_of_skills < len(available_skills):
        #random.sample() does not select the same list element more than once
        selected_skills = random.sample(
            available_skills,
            number_of_skills
        )

    else:

        selected_skills = available_skills.copy()
        remaining_skills = [
            skill
            for skill in technical_skills
            if skill not in selected_skills
        ]

        additional_count = (
            number_of_skills - len(selected_skills)
        )

        """We used extend() instead of append because extend() adds each element individually 
           whereas append() adds the whole object as one element"""
        selected_skills.extend(
            random.sample(
                remaining_skills,
                additional_count
            )
        )
        
    return selected_skills
```

```python
#Generating a random number of projects and select those projects
def generate_projects():

    number_of_projects = random.choices(
        list(project_count_distribution.keys()),
        weights = list(project_count_distribution.values()),
        k = 1
    )[0]

    selected_projects = random.sample(
        projects,
        number_of_projects
    )

    return number_of_projects, selected_projects
```

```python
#Generating a random number of internships and select those internships
def generate_internships():

    number_of_internships = random.choices(
        list(internship_count_distribution.keys()),
        weights = list(internship_count_distribution.values()),
        k = 1
    )[0]

    selected_internships = random.sample(
        internships,
        number_of_internships
    )

    return number_of_internships, selected_internships
```

```python
#Generating a random number of certifications and select those certifications
def generate_certifications():

    number_of_certifications = random.choices(
        list(certification_count_distribution.keys()),
        weights=list(certification_count_distribution.values()),
        k=1
    )[0]

    selected_certifications = random.sample(
        certifications,
        number_of_certifications
    )

    return number_of_certifications, selected_certifications
```

```python
#Generating a random number of achievements and select those achievements
def generate_achievements():

    number_of_achievements = random.choices(
        list(achievement_count_distribution.keys()),
        weights=list(achievement_count_distribution.values()),
        k=1
    )[0]

    selected_achievements = random.sample(
        achievements,
        number_of_achievements
    )

    return number_of_achievements, selected_achievements
```

```python
#Generating the full candidate data
def generate_candidate():
    candidate = {}

    #Degree
    candidate["Degree"] = random.choices(
        list(degrees.keys()),
        weights = list(degrees.values()),
        k = 1
    )[0]

    #Specialization
    candidate["Specialization"] = random.choices(
        list(specializations.keys()),
        weights=list(specializations.values()),
        k=1
    )[0]

    #CGPA
    #np.clip() forces the value to stay between the given minimum and maximum
    cgpa = np.clip(
        np.random.normal(            #Generates a random value from a normal distribution
            loc = 7.7,               #It is the mean
            scale = 0.85             #It is the standard deviation
        ),
        6.0,
        10.0
    )

    candidate["CGPA"] = round(cgpa, 2)

    """Generating Class 10% using a linear relationship with CGPA, where 7.2 is the chosen slope,
       added Gaussian noise with mean 0 and standard deviation 5 to introduce variation,
       and clipped the final value between 60 and 100 to maintain a realistic percentage range"""
    class_10 = np.clip(
        62+
        (cgpa - 5) * 7.2 +
        np.random.normal(0, 5),
        60,
        100
    )

    candidate["Class 10 Percentage"] = round(class_10, 2)

    """Generating Class 12% using a linear relationship with CGPA, where 7.5 is the chosen slope,
       added Gaussian noise with mean 0 and standard deviation 5 to introduce variation,
       and clipped the final value between 60 and 100 to maintain a realistic percentage range"""
    class_12 = np.clip(
        62+
        (cgpa - 5) * 7.5 +
        np.random.normal(0, 5),
        60,
        100
    )

    candidate["Class 12 Percentage"] = round(class_12, 2)

    """I calculate backlog lambda using 1.2 - (cgpa - 6) * 0.35. 
       Here, 6 is the reference CGPA, 1.2 is the baseline expected backlog rate,
       and 0.35 controls how much the expected backlog rate decreases for every one-point increase in CGPA.
       I then clip lambda between 0.05 and 1.5 to keep it valid and reasonable for a Poisson distribution.
       Finally, I use this lambda in np.random.poisson() to generate the actual number of backlogs"""
    backlog_lambda = np.clip(
        1.2 - 
        (cgpa - 6) * 0.35,
        0.05,
        1.5
    )

    candidate["Backlogs"] = min(
        np.random.poisson(backlog_lambda),
        6
    )

    #Skills
    candidate["Skills"] = generate_skills()

    #Projects
    number_of_projects, selected_projects = generate_projects()
    
    candidate["Number of Projects"] = number_of_projects
    candidate["Projects"] = (
        selected_projects
        if number_of_projects > 0
        else "None"
    )

    #Internships
    number_of_internships, selected_internships = generate_internships()

    candidate["Number of Internships"] = number_of_internships
    candidate["Internships"] = (
        selected_internships
        if number_of_internships > 0
        else "None"
    )

    #Certifications
    number_of_certifications, selected_certifications = generate_certifications()

    candidate["Number of Certifications"] = number_of_certifications
    candidate["Certifications"] = (
        selected_certifications
        if number_of_certifications > 0
        else "None"
    )

    #Achievements
    number_of_achievements, selected_achievements = generate_achievements()

    candidate["Number of Achievements"] = number_of_achievements
    candidate["Achievements"] = (
        selected_achievements
        if number_of_achievements > 0
        else "None"
    )
    
    return candidate
```

```python
#Generate 10,000 candidates
candidates = []

for i in range(N):
    
    candidate = generate_candidate()
    candidate["Candidate ID"] = f"CND_{i + 1:05d}"            #Generates Candidate ID
     
    candidates.append(candidate)
```

```python
#Relationship between candidates and jobs
def generate_applications(candidates_df, jobs_df):

    applications = []

    application_id = 1

    for candidate_id in candidates_df["Candidate ID"]:

        number_of_applications = random.choices(
            list(application_count_distribution.keys()),
            weights=list(application_count_distribution.values()),
            k=1
        )[0]

        selected_jobs = random.sample(
            list(jobs_df["Job ID"]),
            number_of_applications
        )

        for job_id in selected_jobs:

            job_role = jobs_df.loc[
                jobs_df["Job ID"] == job_id,
                "Job Role"
            ].values[0]

            applications.append({
                "Application ID": f"APP_{application_id:05d}",
                "Candidate ID": candidate_id,
                "Job ID": job_id,
                "Job Role": job_role
            })

            application_id += 1

    return pd.DataFrame(applications)
```
